```python
# data loader component
# DATASET LOADER CLASS

class FingeringDataset(Dataset):
    def __init__(self, sequences):
        self.sequences = sequences

    def __len__(self):
        return len(self.sequences)

    def __getitem__(self, idx):
        X_list, y_list = self.sequences[idx]
        X_tensor = torch.tensor(X_list, dtype=torch.float32)  # (seq_len, 6)
        y_tensor = torch.tensor(y_list, dtype=torch.long)     # (seq_len,)
        return X_tensor, y_tensor

train_dataset = FingeringDataset(train_seqs)
val_dataset   = FingeringDataset(val_seqs)
test_dataset  = FingeringDataset(test_seqs)

train_loader = DataLoader(train_dataset, batch_size=1, shuffle=True)
val_loader   = DataLoader(val_dataset,   batch_size=1, shuffle=False)
test_loader  = DataLoader(test_dataset,  batch_size=1, shuffle=False)
```

# Modify Architecture and Main Training Loop

```python
from sklearn.metrics import accuracy_score

def train_sequence_model(model, train_loader, val_loader, num_epochs=5, lr=1e-3):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0.0
        for X_seq, y_seq in train_loader:
            X_seq, y_seq = X_seq.to(device), y_seq.to(device)
            optimizer.zero_grad()
            logits, _ = model(X_seq)  # Extract attention later
            logits_reshaped = logits.view(-1, logits.size(-1))
            y_seq_reshaped = y_seq.view(-1)
            loss = criterion(logits_reshaped, y_seq_reshaped)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        avg_loss = total_loss / len(train_loader)
        val_acc = evaluate_sequence_model(model, val_loader)
        print(f"Epoch {epoch+1}/{num_epochs} | Loss: {avg_loss:.4f} | ValAcc: {val_acc:.4f}")

def evaluate_sequence_model(model, data_loader):
    device = next(model.parameters()).device
    model.eval()

    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_seq, y_seq in data_loader:
            X_seq, y_seq = X_seq.to(device), y_seq.to(device)
            logits, _ = model(X_seq)
            preds = torch.argmax(logits, dim=-1)
            all_preds.extend(preds.view(-1).cpu().numpy())
            all_labels.extend(y_seq.view(-1).cpu().numpy())

    return accuracy_score(all_labels, all_preds)
```

```python
import math
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer('pe', pe)

    def forward(self, x):
        seq_len = x.size(0)
        return x + self.pe[:seq_len, :].unsqueeze(1)

class CustomTransformerEncoderLayer(nn.TransformerEncoderLayer):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)

    def forward(self, src, src_mask=None, src_key_padding_mask=None):
        src2, attn_weights = self.self_attn(
            src, src, src,
            attn_mask=src_mask,
            key_padding_mask=src_key_padding_mask,
            need_weights=True,
            average_attn_weights=False
        )
        src = src + self.dropout1(src2)
        src = self.norm1(src)
        src2 = self.linear2(self.dropout(self.activation(self.linear1(src))))
        src = src + self.dropout2(src2)
        src = self.norm2(src)
        return src, attn_weights

class MultiFeatureTransformerTagger(nn.Module):
    def __init__(self, num_pitches, num_fingers, pitch_emb_dim=64, numeric_dim=5, numeric_emb_dim=64,
                 nhead=4, num_layers=2, dim_feedforward=256, dropout=0.1):
        super().__init__()

        self.pitch_embedding = nn.Embedding(num_pitches, pitch_emb_dim)
        self.numeric_projection = nn.Linear(numeric_dim, numeric_emb_dim)
        self.d_model = pitch_emb_dim + numeric_emb_dim
        self.pos_encoder = PositionalEncoding(d_model=self.d_model)

        encoder_layer = CustomTransformerEncoderLayer(
            d_model=self.d_model, nhead=nhead, dim_feedforward=dim_feedforward, dropout=dropout, batch_first=False
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc = nn.Linear(self.d_model, num_fingers)

    def forward(self, x):
        pitch_ids = x[..., 0].long()
        numeric_feats = x[..., 1:].float()
        pitch_emb = self.pitch_embedding(pitch_ids)
        numeric_emb = self.numeric_projection(numeric_feats)
        combined_emb = torch.cat((pitch_emb, numeric_emb), dim=2)
        combined_emb = combined_emb.transpose(0, 1)
        encoded = self.pos_encoder(combined_emb)

        attention_matrices = []
        out = encoded
        for layer in self.transformer_encoder.layers:
            out, attn_weights = layer(out)
            attention_matrices.append(attn_weights)

        logits = self.fc(out)
        logits = logits.transpose(0, 1)
        return logits, attention_matrices
```

```python
###############################################
# Train Transformer
###############################################
#   pitch_vocab = sorted set of pitches
#   finger_vocab = sorted set of finger labels
#   train_loader, val_loader, test_loader
#   Each batch => (X_seq, y_seq) with X_seq.shape=(batch_size, seq_len, 6)

from torch.optim import Adam
import math

transformer_model = MultiFeatureTransformerTagger(
    num_pitches=len(pitch_to_int_mapping),
    num_fingers=len(finger_to_int_mapping),
    pitch_emb_dim=12,       # dimension for pitch embedding
    numeric_dim=5,           # e.g. onset_time, offset_time, onset_vel, offset_vel, channel
    numeric_emb_dim=32,     # project numeric feats to 128
    nhead=2,
    num_layers=2,
    dim_feedforward=128,     # example dimension in feedforward
    dropout=0.05
)

state_dict = torch.load('transformer_model_weights_FINETUNED.pth')
transformer_model.load_state_dict(state_dict)

print("\n=== Training Multi-Feature TRANSFORMER ===")
train_sequence_model(transformer_model, train_loader, val_loader, num_epochs=150, lr=1e-3)
test_acc_transformer = evaluate_sequence_model(transformer_model, test_loader)
print(f"Transformer Test Accuracy = {test_acc_transformer*100:.2f}%")
torch.save(transformer_model.state_dict(), "transformer_model_weights_FINETUNED.pth")    # we save the RNN model weights at the very end so we can load them in later!
```

```python
def predict_piece_fingerings_multi(
    model,
    file_path,
    pitch_to_idx,         # dict mapping spelled pitches -> int
    finger_to_idx=None,   # dict mapping finger labels -> int (optional)
    idx_to_finger=None,   # reverse mapping to decode predictions (optional)
    device=None
):
    df = pd.read_table(file_path, sep="\t", skiprows=1,
        names=[
            "noteID",
            "onset_time",
            "offset_time",
            "spelled_pitch",
            "onset_velocity",
            "offset_velocity",
            "channel",
            "finger_number",
        ],
    )

    # Encode spelled pitch -> int and finger labels -> int
    df["spelled_pitch_int"] = df["spelled_pitch"].map(pitch_to_int_mapping).fillna(0).astype(int)

    if finger_to_idx is not None:
        df["finger_int"] = df["finger_number"].astype(str).map(finger_to_int_mapping).fillna(0).astype(int)
        y_true = df["finger_int"].values
    else:
        y_true = None

    #    Construct multi-feature array X of shape (seq_len, 6)
    #    X[..., 0] = spelled_pitch_int (categorical)
    #    X[..., 1] = onset_time
    #    X[..., 2] = offset_time
    #    X[..., 3] = onset_velocity
    #    X[..., 4] = offset_velocity
    #    X[..., 5] = channel
    seq_len = len(df)
    X_array = np.zeros((seq_len, 6), dtype=np.float32)

    X_array[:, 0] = df["spelled_pitch_int"].astype(float)
    X_array[:, 1] = df["onset_time"].astype(float)
    X_array[:, 2] = df["offset_time"].astype(float)
    X_array[:, 3] = df["onset_velocity"].astype(float)
    X_array[:, 4] = df["offset_velocity"].astype(float)
    X_array[:, 5] = df["channel"].astype(float)

    X_tensor = torch.tensor(X_array, dtype=torch.float32).unsqueeze(0)    # convert X_array into a pytorch tensor

    # move everything to gpu if possible
    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    X_tensor = X_tensor.to(device)
    model = model.to(device)
    model.eval()

    with torch.no_grad():
        logits = model(X_tensor)[0]
        preds = torch.argmax(logits, dim=-1).squeeze(0).cpu().numpy()

    if idx_to_finger:
        y_pred_labels = [int_to_finger_mapping.get(idx, "UNK") for idx in preds]
    else:
        y_pred_labels = preds
    acc = None
    y_true_labels = None
    if y_true is not None:
        from sklearn.metrics import accuracy_score
        acc = accuracy_score(y_true, preds) * 100.0
        if idx_to_finger:
            y_true_labels = [int_to_finger_mapping.get(idx, "UNK") for idx in y_true]

    return preds, y_true, y_pred_labels, y_true_labels, acc
```

# Capturing Attention Weights
A strong “diagonal” means the token mostly attends to itself; off-diagonal structures can show interesting patterns (e.g., focusing on earlier tokens).

```python
import torch
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import os

############################################
# 1) Load and Process the Specific File
############################################
def load_file_to_tensor(file_path, pitch_to_int_mapping, max_seq_len=256):
    """
    Reads a fingering file, encodes the notes into a tensor the Transformer model can process.
    
    Returns:
    - input_tensor: Tensor of shape [1, seq_len, 6] (batch_size=1)
    """
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist.")

    # Read file into DataFrame
    df = pd.read_table(
        file_path, sep="\t", skiprows=1,
        names=["noteID", "onset_time", "offset_time", "spelled_pitch",
               "onset_velocity", "offset_velocity", "channel", "finger_number"]
    )
    
    # Convert spelled pitch to int, defaulting to 0 for unknown pitches
    df["pitch_int"] = df["spelled_pitch"].map(pitch_to_int_mapping).fillna(0).astype(int)
    
    # Build feature array (ensure it doesn't exceed `max_seq_len`)
    seq_len = min(len(df), max_seq_len)
    X_array = np.zeros((seq_len, 6), dtype=np.float32)
    
    X_array[:, 0] = df["pitch_int"][:seq_len].astype(float)
    X_array[:, 1] = df["onset_time"][:seq_len].astype(float)
    X_array[:, 2] = df["offset_time"][:seq_len].astype(float)
    X_array[:, 3] = df["onset_velocity"][:seq_len].astype(float)
    X_array[:, 4] = df["offset_velocity"][:seq_len].astype(float)
    X_array[:, 5] = df["channel"][:seq_len].astype(float)

    # Convert to PyTorch tensor, add batch dimension (1, seq_len, features)
    input_tensor = torch.tensor(X_array, dtype=torch.float32).unsqueeze(0)

    return input_tensor


############################################
# 2) Plot Attention for a Specific Layer & Head
############################################
def plot_attention(attn_matrices, layer=0, head=0, batch_idx=0):
    """
    Visualizes attention heatmap for a given layer/head in the Transformer.
    """
    attn_layer = attn_matrices[layer]  # Extract layer (batch_size, n_heads, seq_len, seq_len)
    attn_map = attn_layer[batch_idx, head].detach().cpu().numpy()

    plt.figure(figsize=(8, 6))
    sns.heatmap(attn_map, cmap="Blues", annot=False)
    plt.xlabel("Key Positions")
    plt.ylabel("Query Positions")
    plt.title(f"Attention Heatmap | Layer {layer}, Head {head}")
    plt.show()
    plt.savefig("ATTENTION_SCORES.png")


############################################
# 3) Attention Rollout / Aggregation
############################################
def attention_rollout(attn_matrices):
    """
    Aggregates attention across layers by multiplying matrices to see cumulative focus.
    """
    batch0 = attn_matrices[0][0]  # Extract first batch
    seq_len = batch0.shape[-1]
    rollout = torch.eye(seq_len, device=batch0.device)  # Identity matrix

    for attn in attn_matrices:
        attn_b0 = attn[0]  # Select batch=0
        attn_mean = attn_b0.mean(dim=0)  # Average across heads
        rollout = attn_mean @ rollout  # Matrix multiplication

    return rollout.detach().cpu().numpy()

def plot_attention_rollout(attn_matrices):
    """
    Plots the aggregated attention across all Transformer layers.
    """
    rollout = attention_rollout(attn_matrices)
    
    plt.figure(figsize=(8, 6))
    sns.heatmap(rollout, cmap="Reds", annot=False)
    plt.xlabel("Key Positions")
    plt.ylabel("Query Positions")
    plt.title("Aggregated Attention Rollout")
    plt.show()
    plt.savefig("ATTENTION_ROLLOUT.png")


############################################
# 4) Load Model and Run Attention Analysis
############################################
transformer_model = MultiFeatureTransformerTagger(
    num_pitches=len(pitch_to_int_mapping),
    num_fingers=len(finger_to_int_mapping),
    pitch_emb_dim=12,
    numeric_dim=5,
    numeric_emb_dim=32,
    nhead=2,
    num_layers=2,
    dim_feedforward=128,
    dropout=0.05
)

state_dict = torch.load('transformer_model_weights_FINETUNED.pth')
transformer_model.load_state_dict(state_dict)
transformer_model.eval()

# Load Specific File
SPECIFIC_FILEPATH = '/home/willy/Desktop/projects/229_piano_fingerings/sequence_models/PianoFingeringDataset_v1.2/FingeringFiles/143-1_fingering.txt'

sample_input = load_file_to_tensor(SPECIFIC_FILEPATH, pitch_to_int_mapping)

# Run through model to get logits and attention
with torch.no_grad():
    logits, attn_matrices = transformer_model(sample_input)

# 1) Plot a single attention layer/head
plot_attention(attn_matrices, layer=1, head=0, batch_idx=0)

# 2) Plot attention rollout
plot_attention_rollout(attn_matrices)
```

```python
import numpy as np

def convert_fingering_to_pygame_format(
    predicted_fingerings, 
    file_path, 
    pitch_to_idx, 
    int_to_finger_mapping
):
    """
    Converts Transformer fingering predictions into a MIDI-like structure for Pygame.
    
    Inputs:
    - predicted_fingerings: List[int], fingering predictions from the model
    - file_path: str, path to the fingering file (for loading note timing info)
    - pitch_to_idx: Dict, mapping spelled pitches to MIDI integers
    - int_to_finger_mapping: Dict, mapping fingering index back to string labels
    
    Returns:
    - pygame_fingering_data: List[Tuple[int, float, float, int]]
        - (MIDI pitch, onset time, duration, predicted fingering)
    """
    
    # Load the fingering file
    df = pd.read_table(
        file_path, 
        sep="\t", 
        skiprows=1,
        names=[
            "noteID", "onset_time", "offset_time", "spelled_pitch",
            "onset_velocity", "offset_velocity", "channel", "finger_number"
        ]
    )
    
    # Ensure we have the same number of predictions as notes
    if len(predicted_fingerings) != len(df):
        print(f"Warning: Mismatch in predicted fingerings ({len(predicted_fingerings)}) vs notes ({len(df)})")

    pygame_fingering_data = []

    for idx, row in df.iterrows():
        spelled_pitch = row["spelled_pitch"]
        onset_time = row["onset_time"]
        offset_time = row["offset_time"]
        duration = offset_time - onset_time

        # Convert spelled pitch to MIDI-like integer
        midi_pitch = pitch_to_idx.get(spelled_pitch, 60)  # Default to Middle C if missing

        # Convert predicted finger index back to actual label
        pred_finger = int_to_finger_mapping.get(predicted_fingerings[idx], "0")

        # Store (MIDI pitch, onset time, duration, predicted fingering)
        pygame_fingering_data.append((midi_pitch, onset_time, duration, pred_finger))

    return pygame_fingering_data



SPECIFIC_FILEPATH = '/home/willy/Desktop/projects/229_piano_fingerings/sequence_models/PianoFingeringDataset_v1.2/FingeringFiles/143-1_fingering.txt'
file_path = SPECIFIC_FILEPATH

transformer_preds, transformer_true, transformer_pred_labels, transformer_true_labels, transformer_acc = predict_piece_fingerings_multi(
    model=transformer_model,
    file_path=file_path,
    pitch_to_idx=pitch_to_int_mapping,
    finger_to_idx=finger_to_int_mapping,
    idx_to_finger=int_to_finger_mapping
)

print(transformer_preds)


# Example Usage:
pygame_fingering_data = convert_fingering_to_pygame_format(
    transformer_preds, 
    file_path, 
    pitch_to_int_mapping, 
    int_to_finger_mapping
)

print(pygame_fingering_data[:10])  # Print first 10 notes to verify
```

```python
import pygame
import time

# 1) Initialize Pygame
pygame.init()
WIDTH, HEIGHT = 800, 300
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Piano Fingering Playback")

# 2) Define white and black keys
white_keys = [pygame.Rect(x * 40, 0, 40, 200) for x in range(21)]  # 21 white keys (partial keyboard)
black_keys = [pygame.Rect(x * 40 + 30, 0, 30, 120) for x in range(15)]  # 15 black keys (approximation)

# 3) Function to render piano with highlights
def draw_piano(correct_notes, incorrect_notes):
    screen.fill((255, 255, 255))  # White background

    # Draw white keys
    for key in white_keys:
        pygame.draw.rect(screen, (255, 255, 255), key, 0)
        pygame.draw.rect(screen, (0, 0, 0), key, 2)

    # Draw black keys
    for key in black_keys:
        pygame.draw.rect(screen, (0, 0, 0), key, 0)

    # Highlight correct notes (Green)
    for note in correct_notes:
        if note % 21 < len(white_keys):  # Ensure it's within range
            pygame.draw.rect(screen, (0, 255, 0), white_keys[note % 21], 0)

    # Highlight incorrect notes (Red)
    for note in incorrect_notes:
        if note % 21 < len(white_keys):  # Ensure it's within range
            pygame.draw.rect(screen, (255, 0, 0), white_keys[note % 21], 0)

    pygame.display.flip()

# 4) Define fingering playback sequence
fingering_sequence = [
    ([40, 42], [41]),    # Frame 1: Correct = 40, 42 | Incorrect = 41
    ([44, 47], [46]),    # Frame 2: Correct = 44, 47 | Incorrect = 46
    ([50, 52], [51]),    # Frame 3: Correct = 50, 52 | Incorrect = 51
    ([55, 57], [56]),    # Frame 4: Correct = 55, 57 | Incorrect = 56
    ([60, 62], [61]),    # Frame 5: Correct = 60, 62 | Incorrect = 61
]

# 5) Run the playback animation
running = True
for correct_fingering, incorrect_fingering in fingering_sequence:
    draw_piano(correct_fingering, incorrect_fingering)
    time.sleep(2)  # Pause for 2 seconds between frames

pygame.quit()
```

```python
import os
import time
import numpy as np
import pandas as pd
import torch
import cv2

##############################################
# (1) Load Predictions from Transformer Model
##############################################

# Specify your Transformer model and mapping functions
SPECIFIC_FILEPATH = "/home/willy/Desktop/projects/229_piano_fingerings/sequence_models/PianoFingeringDataset_v1.2/FingeringFiles/143-1_fingering.txt"

# Run prediction
transformer_preds, transformer_true, transformer_pred_labels, transformer_true_labels, acc = predict_piece_fingerings_multi(
    model=transformer_model,
    file_path=SPECIFIC_FILEPATH,
    pitch_to_idx=pitch_to_int_mapping,
    finger_to_idx=finger_to_int_mapping,
    idx_to_finger=int_to_finger_mapping
)

# Load the corresponding MIDI file for reference
df = pd.read_table(
    SPECIFIC_FILEPATH,
    sep="\t",
    skiprows=1,
    names=[
        "noteID", "onset_time", "offset_time", "spelled_pitch",
        "onset_velocity", "offset_velocity", "channel", "finger_number"
    ]
)

# Convert spelled pitch to MIDI note number
def spelled_pitch_to_midi(spelled):
    return pitch_to_int_mapping.get(spelled, 60)  # Default to middle C if not found

# Build fingering sequence with correct/incorrect notes and finger labels
fingering_sequence = []
for i in range(len(df)):
    spelled_pitch = df.iloc[i]["spelled_pitch"]
    midi_pitch = spelled_pitch_to_midi(spelled_pitch)
    true_finger = transformer_true_labels[i] if i < len(transformer_true_labels) else "?"
    pred_finger = transformer_pred_labels[i] if i < len(transformer_pred_labels) else "?"

    # Check if prediction is correct
    is_correct = (transformer_preds[i] == transformer_true[i]) if transformer_true is not None and i < len(transformer_true) else True

    if is_correct:
        fingering_sequence.append(([midi_pitch], [true_finger], [], []))  # Green note + correct fingering
    else:
        fingering_sequence.append(([], [], [midi_pitch], [pred_finger]))  # Red note + incorrect fingering


##############################################
# (2) Video Rendering Settings
##############################################

# Video Settings
WIDTH, HEIGHT = 800, 300
FPS = 2  # 2 frames per second (each note stays for 0.5s)
VIDEO_FILENAME = "piano_fingering_visualization.mp4"

# OpenCV Video Writer
fourcc = cv2.VideoWriter_fourcc(*"mp4v")  # MP4 codec
video_out = cv2.VideoWriter(VIDEO_FILENAME, fourcc, FPS, (WIDTH, HEIGHT))

# Define piano layout
WHITE_KEY_WIDTH = WIDTH // 21
WHITE_KEY_HEIGHT = 200
BLACK_KEY_WIDTH = WHITE_KEY_WIDTH - 10
BLACK_KEY_HEIGHT = 120

# Define colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 255, 0)
RED = (255, 0, 0)


##############################################
# (3) Draw & Save Each Frame as Video
##############################################

def draw_piano_frame(correct_notes, correct_fingers, incorrect_notes, incorrect_fingers):
    """
    Generates a single frame where correct notes are highlighted in green,
    incorrect notes are highlighted in red, and finger annotations are displayed.
    """
    frame = np.ones((HEIGHT, WIDTH, 3), dtype=np.uint8) * 255  # White background

    # Draw white keys
    for i in range(21):
        x = i * WHITE_KEY_WIDTH
        cv2.rectangle(frame, (x, 0), (x + WHITE_KEY_WIDTH, WHITE_KEY_HEIGHT), BLACK, 2)

    # Draw black keys (approximate layout for a 21-key section)
    black_key_positions = [1, 3, 6, 8, 10, 13, 15, 18, 20]
    for i in black_key_positions:
        x = i * WHITE_KEY_WIDTH - BLACK_KEY_WIDTH // 2
        cv2.rectangle(frame, (x, 0), (x + BLACK_KEY_WIDTH, BLACK_KEY_HEIGHT), BLACK, -1)

    # Highlight correct notes (green)
    for note in correct_notes:
        key_pos = note % 21
        if 0 <= key_pos < 21:
            x = key_pos * WHITE_KEY_WIDTH
            cv2.rectangle(frame, (x, 0), (x + WHITE_KEY_WIDTH, WHITE_KEY_HEIGHT), GREEN, -1)

    # Highlight incorrect notes (red)
    for note in incorrect_notes:
        key_pos = note % 21
        if 0 <= key_pos < 21:
            x = key_pos * WHITE_KEY_WIDTH
            cv2.rectangle(frame, (x, 0), (x + WHITE_KEY_WIDTH, WHITE_KEY_HEIGHT), RED, -1)

    # Ensure text is drawn **AFTER** the notes so it stays in the foreground
    for i, note in enumerate(correct_notes):
        key_pos = note % 21
        if 0 <= key_pos < 21:
            x = key_pos * WHITE_KEY_WIDTH
            cv2.putText(frame, str(correct_fingers[i]), (x + 10, WHITE_KEY_HEIGHT - 20),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, BLACK, 20, cv2.LINE_AA)

    for i, note in enumerate(incorrect_notes):
        key_pos = note % 21
        if 0 <= key_pos < 21:
            x = key_pos * WHITE_KEY_WIDTH
            cv2.putText(frame, str(incorrect_fingers[i]), (x + 10, WHITE_KEY_HEIGHT - 50),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, BLACK, 9, cv2.LINE_AA)

    return frame


# Generate video frames
for correct_notes, correct_fingers, incorrect_notes, incorrect_fingers in fingering_sequence:
    frame = draw_piano_frame(correct_notes, correct_fingers, incorrect_notes, incorrect_fingers)
    video_out.write(frame)

# Release video writer
video_out.release()
print(f"Video saved as: {VIDEO_FILENAME}")
```
